```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
df = pd.read_csv('../data/clean/imdb_movies_clean.csv')
df.head()
```

```python
engine = create_engine(
    "mysql+pymysql://root:admin%40123@localhost/imdb_etl"
)
```

```python
df.to_sql(
    name='imdb_movies',
    con=engine,
    if_exists='append',
    index=False
)

print("Data Loaded Successfully")
```

Validation Queries

Total Records

```python
query = "SELECT COUNT(*) AS total_records FROM imdb_movies"

pd.read_sql(query, engine)
```

Check Movie ID Nulls

```python
query = " SELECT COUNT(*) AS missing_movie_id FROM imdb_movies WHERE movie_id IS NULL "
pd.read_sql(query, engine)
```

Check Rating Nulls

```python
query = " SELECT COUNT(*) AS missing_rating FROM imdb_movies WHERE rating IS NULL"

pd.read_sql(query, engine)
```

View Sample Data

```python
query = "SELECT * FROM imdb_movies LIMIT 10"

pd.read_sql(query, engine)
```
